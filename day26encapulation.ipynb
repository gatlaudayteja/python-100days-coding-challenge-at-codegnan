```python
class Bank:
    bank_name = "Codegnan Bank"

    def __init__(self, username, password, amount):
        self.username = username          # public
        self._password = password         # protected
        self.__amount = amount             # private

    def balance(self):
        return f"Current balance is {self.__amount}"

    def withdraw(self, withdraw_amount):
        if self.__amount >= withdraw_amount:
            self.__amount -= withdraw_amount
            return f"{withdraw_amount} withdrawal successful and Current Balance is {self.__amount}"
        return "Insufficient amount"

    def deposit(self, deposit_amount):
        self.__amount += deposit_amount
        return f"{deposit_amount} is added successfully and Current Balance is {self.__amount}"

    def __total_users(self):   # private method
        return "Total users are 98"


# object creation
user = Bank("teja123", "teja12@gmail.com", 2000)

print(user.balance())
print(user.withdraw(500))
print(user._password)          # protected access

# name mangling (accessing private variable)
print(user._Bank__amount)
```
